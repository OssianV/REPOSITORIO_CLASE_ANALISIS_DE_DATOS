# Pandas sorting practice

In this notebook I will practice sorting tasks in Pandas, from the usual cases to more demanding wrangling scenarios. I will focus on the following practice areas:

1. Sorting by values
    - 1.1. Sort a Series or DataFrame by one column
    - 1.2. Sort ascending vs descending
    - 1.3. Sort by multiple columns with different sort directions
    - 1.4. Sort with missing values first or last
    - 1.5. Sort columns instead of rows

2. Sorting by index
    - 2.1. Sort a simple row index
    - 2.2. Sort a column index
    - 2.3. Sort a MultiIndex by one or more levels
    - 2.4. Sort only a chosen level while keeping the others meaningful

3. Multi-key and tie-breaking scenarios
    - 3.1. Break ties with secondary and tertiary columns
    - 3.2. Keep a stable order when duplicate sort keys exist
    - 3.3. Sort and then keep first or last record per entity

4. Specialized value-based sorting
    - 4.1. Get top N values
    - 4.2. Get bottom N values
    - 4.3. Sort by absolute value or distance from a target
    - 4.4. Sort by a derived metric built from multiple columns

5. Custom order sorting
    - 5.1. Sort weekdays, months, status labels, or severity levels in business order
    - 5.2. Sort with ordered categoricals
    - 5.3. Sort with a custom mapping dictionary
    - 5.4. Sort text case-insensitively or by string length

6. Text and mixed-type sorting
    - 6.1. Sort numeric-looking strings as numbers instead of text
    - 6.2. Sort strings with prefixes or suffixes after extracting the numeric part
    - 6.3. Sort messy mixed columns after cleaning types
    - 6.4. Compare lexicographic vs natural sort behavior

7. Time-aware sorting
    - 7.1. Sort timestamps chronologically
    - 7.2. Sort after converting strings to datetimes
    - 7.3. Sort latest or earliest record per person, account, or product
    - 7.4. Sort month or weekday names in calendar order instead of alphabetic order

8. Group-wise sorting tasks
    - 8.1. Sort rows within each group
    - 8.2. Get top N rows per group
    - 8.3. Rank rows inside each group after sorting
    - 8.4. Sort aggregated groupby results by one or more metrics

9. Reshaped and hierarchical data sorting
    - 9.1. Sort after pivot, pivot_table, stack, or unstack
    - 9.2. Sort hierarchical rows and columns produced by aggregation
    - 9.3. Sort after melt or explode to restore a useful reading order

10. Sorting as part of a wrangling workflow
    - 10.1. Sort before dropping duplicates to keep the right row
    - 10.2. Sort before cumulative operations like cumsum or diff
    - 10.3. Sort before merge_asof or similar ordered joins
    - 10.4. Sort for reporting after filtering and feature engineering

11. Performance and robustness practice
    - 11.1. Choose between sort_values, nlargest, and nsmallest
    - 11.2. Practice sorting large data with only needed columns
    - 11.3. Validate that the result is really sorted as expected
    - 11.4. Watch for hidden issues: NaNs, duplicates, mixed dtypes, and object columns

The idea is to work with smallish datasets (200 MB or less), that are either in an `.xlsx`, `.csv`, or some variant of a sql file (like `sqlite` and others). To get practice with excel basics (which i think i am already confident at), i will perform the same sorting exercises in excel.
